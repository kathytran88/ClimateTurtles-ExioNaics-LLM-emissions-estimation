```python
from sentence_transformers import SentenceTransformer 
import torch 
import torch.nn as nn 
```

```python
def loss_mnr (z1,z2,temperature=0.07):
    z1=F.normalize(z1,dim=1)
    z2=F.normalize(z2,dim=1)
    sim_matrix=(z1@z2.T)/temperature 
    diagonal_labels=torch.arange(z1.size(0),device=z1.device) 
    z1_loss=F.cross_entropy(sim_matrix,diagonal_labels)
    z2_loss=F.cross_entropy(sim_matrix.T,diagonal_labels)
    total_loss=(z1_loss+z2_loss)/2
    return total_loss
```

```python
class SBERTClassifier(nn.Module):
    def __init__(self,model_sbert,num_classes):
        super().__init__()
        self.pretrained=model_sbert
        self.emb_dim=model_sbert.get_sentence_embedding_dimension()
        self.projector=nn.Linear(self.emb_dim,128)
        self.classifier=nn.Linear(self.emb_dim,num_classes)
    
    def forward(self,texts):
        embedding=self.pretrained.encode(texts,convert_to_tensor=True,device=device).clone()
        z1=self.projector(embedding)
        z2=self.projector(embedding)
        logit=self.classifier(embedding)
        return z1,z2,logit 
```

```python
# try an unfrozen backbone implementation 
from sentence_transformers import util 

def loss_mnr2 (z1,z2,temperature=0.1): # change temperature to be more balanced 
    z1=F.normalize(z1,dim=1)
    z2=F.normalize(z2,dim=1)
    sim_matrix=(z1@z2.T)/temperature 
    diagonal_labels=torch.arange(z1.size(0),device=z1.device) 
    z1_loss=F.cross_entropy(sim_matrix,diagonal_labels)
    z2_loss=F.cross_entropy(sim_matrix.T,diagonal_labels)
    total_loss=(z1_loss+z2_loss)/2
    return total_loss

class SBERTClassifier2(nn.Module):
    def __init__(self,model_sbert,num_classes):
        super().__init__()
        self.pretrained=model_sbert
        self.emb_dim=model_sbert.get_sentence_embedding_dimension()
        self.projector=nn.Linear(self.emb_dim,128)
        self.classifier=nn.Linear(self.emb_dim,num_classes)
    
    def forward(self,texts):
        encoded=self.pretrained.tokenize(texts)
        encoded=util.batch_to_device(encoded,device)
        output=self.pretrained(encoded)
        embedding=output.get('sentence_embedding')

        z1=self.projector(embedding)
        z2=self.projector(embedding)
        logit=self.classifier(embedding)
        return z1,z2,logit 

num_classes=df['NAICS_2 Title'].nunique()
pretrained=SentenceTransformer("all-MiniLM-L6-v2")
model=SBERTClassifier2(pretrained,num_classes)
device=torch.device('cpu')
model.to(device)
```

```python
from transformers import get_linear_schedule_with_warmup 
epochs_num=20
steps_total=len(loader_train)*epochs_num
warmups_total=int(0.1*steps_total) # bring back the warm ups since we are doing the unfrozen backbone from scratch (+ will do early stopping this time)

optimizer=torch.optim.AdamW([{'params':model.pretrained.parameters(),'lr':2e-6},{'params':model.projector.parameters(),'lr':2e-5},
{'params':model.classifier.parameters(),'lr':2e-5},]) #smaller LR for backbone as it is unfrozen in this case 

schedule=get_linear_schedule_with_warmup(optimizer,num_warmup_steps=warmups_total,num_training_steps=steps_total)

acc1_best_val=0
zero_improvement=0

for one_epoch in range (epochs_num):
    model.train()
    loss_total=0
    for batch in loader_train:  
        company_description=batch[0]
        naics2_labels=batch[2].to(device)
        naics2_description=batch[1]

        z1,_,_=model(company_description)
        z2,_,logit=model(naics2_description)
        contrast_loss=loss_mnr2(z1,z2)
        classifi_loss=F.cross_entropy(logit,naics2_labels,label_smoothing=0.05)
        combine_loss=contrast_loss+classifi_loss
        combine_loss.backward()
        optimizer.step()
        schedule.step()

        loss_total+=combine_loss.item()
        current_loss=loss_total/len(loader_train)
    
    acc1_current_val=topk_accuracy(loader_val)[0]
    if acc1_current_val > acc1_best_val:
        acc1_best_val=acc1_current_val
        zero_improvement=0
        torch.save(model.state_dict(),'ClimateTurtles-ExioNaics-LLM-emissions-estimation/sbert_best_unfrozen.pt')
        print(f"Total epoch {one_epoch+1}, loss {current_loss}, new best acc1 for val {acc1_best_val}")
    
    else: 
        zero_improvement+=1
        if zero_improvement>=3: # wait for 3 epochs when there are zero improvements
            print("Early Stopping for Best ACC1")
            break 
```

```python
def accuracy_topk (logit,labels,k):
    pred_classes_topk=logit.argsort(dim=1,descending=True)[:,:k]
    correct_bool=torch.eq(pred_classes_topk,labels.view(-1,1).expand(-1,k))
    return torch.any(correct_bool,dim=1).float().sum().item()
```

```python
def topk_accuracy(loader):
    model.eval()
    acc1,acc3,acc5,acc10=0,0,0,0
    total=0

    with torch.no_grad():
        for company_description, naics2_description, label in loader: 
            label=torch.tensor(list(label)).to(device)
            logit=model(company_description)[2]
            acc1+=accuracy_topk(logit,label,k=1)
            acc3+=accuracy_topk(logit,label,k=3)
            acc5+=accuracy_topk(logit,label,k=5)
            acc10+=accuracy_topk(logit,label,k=10)
            total+=len(label)
    acc_1=acc1/total
    acc_3=acc3/total
    acc_5=acc5/total
    acc_10=acc10/total


    print("ACC@1:",acc_1)
    print("ACC@3:",acc_3)
    print("ACC@5:",acc_5)
    print("ACC@10:",acc_10)

    return acc_1,acc_3,acc_5,acc_10
```

```python
#hyperparameter tuning (done and fixed above) --> cross validation --> then do test accuracy 
best_acc1=0
results_lst=[]
for learning_rates in [1e-5,3e-5,5e-5]:
    print(f"Current training in {learning_rates} lr")
    pretrained=SentenceTransformer("all-MiniLM-L6-v2")
    model=SBERTClassifier(pretrained,num_classes)
    model.to(device)
    optimizer=torch.optim.AdamW(model.parameters(),lr=learning_rates) #switched from sgd to adamW

    for epoch in range (5):
        model.train()
        loss_total=0

        for company_description,naics2_description, naics2_labels in loader_train: 
            naics2_labels=torch.tensor(list(naics2_labels)).to(device)
            optimizer.zero_grad()
            z1,_,_=model(company_description)
            z2,_,logit=model(naics2_description)
            contrast_loss=loss_mnr(z1,z2)
            classifi_loss=F.cross_entropy(logit,naics2_labels,weight=normalized_weights,label_smoothing=0.1)
            combine_loss=contrast_loss+classifi_loss
            combine_loss.backward()
            optimizer.step()
            loss_total+=combine_loss.item()
        print(f"current epoch{epoch+1},loss:{loss_total/len(loader_train)}") # fixed upon proofreading, but not run again to not interfere with the model with best hyperparameters

    accuracy_val1,accuracy_val3,accuracy_val5,accuracy_val10=topk_accuracy(loader_val)
    results_lst.append((learning_rates,accuracy_val1))
    print(f"When learning rate is {learning_rates}, ACC@1 for validation set is {accuracy_val1}")

    if accuracy_val1>best_acc1:
        best_acc1=accuracy_val1
        torch.save(model.state_dict(),'sbert_hyperparam_best.pt')
        print(f"current best sbert model saved")

print(max(results_lst,key=lambda x: x[1]),"is the best learning rate and best ACC@1 for Validation Set")
       
```

Note: The losses above are very large because they were previously not divided by len(loader_train), which account for the large numbers observed. When the losses for each epoch are divded by len(loader_train), then they are quite small. 

```python
# as the models above with adamW optimizer were less effective than with SGD optimizer with lr=2e-05
# I will now do lr=1e-05 (best performing out of the 3 lrs) with SGD optimizer compare SGD optimizer lr=2e-05 (original)
pretrained=SentenceTransformer("all-MiniLM-L6-v2")
model=SBERTClassifier(pretrained,num_classes)
model.to(device)
optimizer=torch.optim.SGD(model.parameters(),lr=1e-05,momentum=0.85) #switched back to SGD

for epoch in range (5):
    model.train()
    loss_total=0

    for company_description,naics2_description, naics2_labels in loader_train: 
        naics2_labels=torch.tensor(naics2_labels).to(device)
        optimizer.zero_grad()
        z1,_,_=model(company_description)
        z2,_,logit=model(naics2_description)
        contrast_loss=loss_mnr(z1,z2)
        classifi_loss=F.cross_entropy(logit,naics2_labels,label_smoothing=0.1) # no class weights (like original) to see if it improves 
        combine_loss=contrast_loss+classifi_loss
        combine_loss.backward()
        optimizer.step()
        loss_total+=combine_loss.item()
```

```python
from sklearn.metrics import mean_absolute_error
import numpy as np

def topk_accuracy_with_ghg(loader):

    model.eval()
    acc1,acc3,acc5,acc10=0,0,0,0
    total=0
    ghg_predict_acc1=np.array([])
    ghg_predict_acc3=np.array([])
    ghg_predict_acc5=np.array([])
    ghg_predict_acc10=np.array([])
    ghg_per_naics2=df.groupby('NAICS_2 Label')['Carbon Intensity'].mean().to_dict()

    carbon_predict=np.array([])
    truth_ghg_company=np.array([])

    with torch.no_grad():
        for company_description, naics2_description, label, company_revenue, co2 in loader: 
            label=torch.tensor(list(label)).to(device)
            company_revenue=torch.tensor(list(company_revenue),dtype=torch.float)
            co2=torch.tensor(list(co2),dtype=torch.float)
            logit=model(company_description)[2]

            pred_top10=logit.topk(10,dim=1).indices.cpu().numpy() 

            def pred_co2 (k): 
                return torch.tensor([np.mean([ghg_per_naics2[int(naics)]for naics in top10[:k]])for top10 in pred_top10]
                ,dtype=torch.float)*company_revenue*1000000

            ghg_predict_acc1=np.concatenate([ghg_predict_acc1,pred_co2(1).numpy()])
            ghg_predict_acc3=np.concatenate([ghg_predict_acc3,pred_co2(3).numpy()])
            ghg_predict_acc5=np.concatenate([ghg_predict_acc5,pred_co2(5).numpy()])
            ghg_predict_acc10=np.concatenate([ghg_predict_acc10,pred_co2(10).numpy()])

            truth_ghg_company=np.concatenate([truth_ghg_company,co2.numpy()])

            acc1+=accuracy_topk(logit,label,k=1)
            acc3+=accuracy_topk(logit,label,k=3)
            acc5+=accuracy_topk(logit,label,k=5)
            acc10+=accuracy_topk(logit,label,k=10)
            total+=len(label)
    acc_1=acc1/total
    acc_3=acc3/total
    acc_5=acc5/total
    acc_10=acc10/total
    def rmse_and_mae (company_ghg_pred):
        rmse=np.sqrt(np.mean((company_ghg_pred-truth_ghg_company)**2))
        mean_abs_error=mean_absolute_error(truth_ghg_company,company_ghg_pred)
        return rmse, mean_abs_error

    rmse_acc1,mae_acc1=rmse_and_mae(ghg_predict_acc1)
    rmse_acc3,mae_acc3=rmse_and_mae(ghg_predict_acc3)
    rmse_acc5,mae_acc5=rmse_and_mae(ghg_predict_acc5)
    rmse_acc10,mae_acc10=rmse_and_mae(ghg_predict_acc10)

    print("Mean ABS Error for GHG avg by top 1 NAICS2 predictions :",mae_acc1)
    print("RMSE for GHG by top 1 NAICS2 predictions:", rmse_acc1)

    print("Mean ABS Error for GHG avg by top 3 NAICS2 predictions :",mae_acc3)
    print("RMSE for GHG by top 3 NAICS2 predictions:", rmse_acc3)

    print("Mean ABS Error for GHG avg by top 5 NAICS2 predictions :",mae_acc5)
    print("RMSE for GHG by top 5 NAICS2 predictions:", rmse_acc5)

    print("Mean ABS Error for GHG avg by top 10 NAICS2 predictions :",mae_acc10)
    print("RMSE for GHG by top 10 NAICS2 predictions:", rmse_acc10)

    print("ACC@1 for NAICS2 Classification:",acc_1)
    print("ACC@3 for NAICS2 Classification:",acc_3)
    print("ACC@5 for NAICS2 Classification:",acc_5)
    print("ACC@10 for NAICS2 Classification:",acc_10)

    return acc_1,acc_3,acc_5,acc_10, rmse_acc1,mae_acc1, rmse_acc3,mae_acc3, rmse_acc5,mae_acc5, rmse_acc10,mae_acc10
```
