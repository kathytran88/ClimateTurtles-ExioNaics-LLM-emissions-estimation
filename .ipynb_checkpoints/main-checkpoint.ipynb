```python
# fine tune sentence bert model to predict naics 2 code given enterprise description then get accuracy, recall, precision scores
from sentence_transformers import SentenceTransformer 
import torch 
import torch.nn as nn 
```

```python
def loss_simcse (z1,z2,temperature=0.05):
    z1=F.normalize(z1,dim=1)
    z2=F.normalize(z2,dim=1)
    sim_matrix=F.cosine_similarity(z1.unsqueeze(1),z2.unsqueeze(0),dim=2)/temperature
    naics2_labels=torch.tensor(range(z1.size(0)),device=z1.device) 
    return F.cross_entropy(sim_matrix,naics2_labels)
```

```python
class SBERTClassifier(nn.Module):
    def __init__(self,model_sbert,num_classes):
        super().__init__()
        self.pretrained=model_sbert
        self.emb_dim=model_sbert.get_sentence_embedding_dimension()
        self.projector=nn.Linear(self.emb_dim,128)
        self.classifier=nn.Linear(self.emb_dim,num_classes)
    
    def forward(self,texts):
        embedding=self.pretrained.encode(texts,convert_to_tensor=True,device=device).clone()
        z1=self.projector(embedding)
        z2=self.projector(embedding)
        logit=self.classifier(embedding)
        return z1,z2,logit 
```

```python
num_classes=df['NAICS_2 Code'].nunique()
pretrained=SentenceTransformer("all-MiniLM-L6-v2")
model=SBERTClassifier(pretrained,num_classes)
```

```python
device=torch.device('cpu')
model.to(device)
```

```python
optimizer=torch.optim.SGD(model.parameters(),lr=2e-5,momentum=0.85)
criterion=nn.CrossEntropyLoss() # classifying companies into NAICS2 code
```

```python
from sklearn.preprocessing import LabelEncoder
encoder=LabelEncoder()
df['NAICS_2 Label']=encoder.fit_transform(df['NAICS_2 Title'])
train['NAICS_2 Label']=encoder.transform(train['NAICS_2 Title'])
val['NAICS_2 Label']=encoder.transform(val['NAICS_2 Title'])
test['NAICS_2 Label']=encoder.transform(test['NAICS_2 Title'])

from torch.utils.data import DataLoader 
loader_train=DataLoader(list(zip(train['Description'].tolist(),train['NAICS_2 Label'].tolist())),batch_size=32,shuffle=True)
loader_val=DataLoader(list(zip(val['Description'].tolist(),val['NAICS_2 Label'].tolist())),batch_size=32)
loader_test=DataLoader(list(zip(test['Description'].tolist(),test['NAICS_2 Label'].tolist())),batch_size=32)
```

```python
import torch.nn.functional as F 

for epoch in range (10):
    model.train()
    loss_total=0

    for description,label in loader_train: 
        label=torch.tensor(label).to(device)
        optimizer.zero_grad()
        z1,z2,logit=model(description)
        contrast_loss=loss_simcse(z1,z2)
        classifi_loss=F.cross_entropy(logit,label,label_smoothing=0.1)
        combine_loss=contrast_loss+classifi_loss
        combine_loss.backward()
        optimizer.step()
        loss_total+=combine_loss.item()

# need to change the loss function during contrastive learning to multiple negatives ranking loss instead of simCSE
# need to check if i need to rerun sbert training everytime (although saved)
# sbert outputs an embeding! (Double check this) --> Tomas suggested putting the outputs through a multiclass logistic regression 
# need to check about the text pairings 
# need to do validation 
```
