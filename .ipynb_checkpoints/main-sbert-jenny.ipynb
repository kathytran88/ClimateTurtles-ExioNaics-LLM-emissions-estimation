```python
from sentence_transformers import SentenceTransformer 
import torch 
import torch.nn as nn 
```

```python
def loss_mnr (z1,z2,temperature=0.05):
    z1=F.normalize(z1,dim=1)
    z2=F.normalize(z2,dim=1)
    sim_matrix=(z1@z2.T)/temperature 
    naics2_labels=torch.arange(z1.size(0),device=z1.device) 
    z1_loss=F.cross_entropy(sim_matrix,naics2_labels)
    z2_loss=F.cross_entropy(sim_matrix.T,naics2_labels)
    total_loss=(z1_loss+z2_loss)/2
    return total_loss
```

```python
class SBERTClassifier(nn.Module):
    def __init__(self,model_sbert,num_classes):
        super().__init__()
        self.pretrained=model_sbert
        self.emb_dim=model_sbert.get_sentence_embedding_dimension()
        self.projector=nn.Linear(self.emb_dim,128)
        self.classifier=nn.Linear(self.emb_dim,num_classes)
    
    def forward(self,texts):
        embedding=self.pretrained.encode(texts,convert_to_tensor=True,device=device).clone()
        z1=self.projector(embedding)
        z2=self.projector(embedding)
        logit=self.classifier(embedding)
        return z1,z2,logit 
```

```python
from sklearn.preprocessing import LabelEncoder
encoder=LabelEncoder()
df['NAICS_2 Label']=encoder.fit_transform(df['NAICS_2 Title'])
train['NAICS_2 Label']=encoder.transform(train['NAICS_2 Title'])
val['NAICS_2 Label']=encoder.transform(val['NAICS_2 Title'])
test['NAICS_2 Label']=encoder.transform(test['NAICS_2 Title'])

from torch.utils.data import DataLoader 
loader_train=DataLoader(list(zip(train['Company Description'].tolist(),train['NAICS_2 Description'].tolist(),
train['NAICS_2 Label'].tolist())),batch_size=32,shuffle=True,drop_last=True)
loader_val=DataLoader(list(zip(val['Company Description'].tolist(),val['NAICS_2 Description'].tolist(),
val['NAICS_2 Label'].tolist())),batch_size=32,drop_last=True)
loader_test=DataLoader(list(zip(test['Company Description'].tolist(),test['NAICS_2 Description'].tolist(),
test['NAICS_2 Label'].tolist())),batch_size=32,drop_last=True)
```

```python
num_classes=df['NAICS_2 Title'].nunique()
pretrained=SentenceTransformer("all-MiniLM-L6-v2")
model=SBERTClassifier(pretrained,num_classes)
device=torch.device('cpu')
model.to(device)
optimizer=torch.optim.SGD(model.parameters(),lr=2e-5,momentum=0.85)
criterion=nn.CrossEntropyLoss() # classifying companies into NAICS2 code
```

```python
import torch.nn.functional as F 

for epoch in range (5):
    model.train()
    loss_total=0

    for company_description,naics2_description, naics2_labels in loader_train: 
        naics2_labels=torch.tensor(naics2_labels).to(device)
        optimizer.zero_grad()
        z1,_,_=model(company_description)
        z2,_,logit=model(naics2_description)
        contrast_loss=loss_mnr(z1,z2)
        classifi_loss=F.cross_entropy(logit,naics2_labels,label_smoothing=0.1)
        combine_loss=contrast_loss+classifi_loss
        combine_loss.backward()
        optimizer.step()
        loss_total+=combine_loss.item()
```

```python
# torch.save(model.state_dict(),'sbert_5epochs_32batch.pt') 
pretrained=SentenceTransformer("all-MiniLM-L6-v2")
model=SBERTClassifier(pretrained,num_classes)
model.load_state_dict(torch.load('sbert_5epochs_32batch.pt'))
model.to(device)
model.eval() 
```

```python
def accuracy_topk (logit,labels,k):
    pred_classes_topk=logit.argsort(dim=1,descending=True)[:,:k]
    correct_bool=torch.eq(pred_classes_topk,labels.view(-1,1).expand(-1,k))
    return torch.any(correct_bool,dim=1).float().sum().item()
```

```python
def topk_accuracy(loader):
    model.eval()
    acc1,acc3,acc5,acc10=0,0,0,0
    total=0

    with torch.no_grad():
        for company_description, naics2_description, label in loader: 
            label=torch.tensor(list(label)).to(device)
            logit=model(company_description)[2]
            acc1+=accuracy_topk(logit,label,k=1)
            acc3+=accuracy_topk(logit,label,k=3)
            acc5+=accuracy_topk(logit,label,k=5)
            acc10+=accuracy_topk(logit,label,k=10)
            total+=len(label)
    acc_1=acc1/total
    acc_3=acc3/total
    acc_5=acc5/total
    acc_10=acc10/total


    print("ACC@1:",acc_1)
    print("ACC@3:",acc_3)
    print("ACC@5:",acc_5)
    print("ACC@10:",acc_10)

    return acc_1,acc_3,acc_5,acc_10
```

```python
from sklearn.utils.class_weight import compute_class_weight
class_counts=train['NAICS_2 Label'].value_counts().sort_index()
normalized_weights=torch.tensor(1.0/class_counts.values,dtype=torch.float).to(device)
classifi_loss=F.cross_entropy(logit,naics2_labels,weight=normalized_weights,label_smoothing=0.1)
```

```python
#hyperparameter tuning --> cross validation --> then do test accuracy 
best_acc1=0
results_lst=[]
for learning_rates in [1e-5,3e-5,5e-5]:
    print(f"Current training in {learning_rates} lr")
    pretrained=SentenceTransformer("all-MiniLM-L6-v2")
    model=SBERTClassifier(pretrained,num_classes)
    model.to(device)
    optimizer=torch.optim.AdamW(model.parameters(),lr=learning_rates) #switched from sgd to adamW

    for epoch in range (5):
        model.train()
        loss_total=0

        for company_description,naics2_description, naics2_labels in loader_train: 
            naics2_labels=torch.tensor(list(naics2_labels)).to(device)
            optimizer.zero_grad()
            z1,_,_=model(company_description)
            z2,_,logit=model(naics2_description)
            contrast_loss=loss_mnr(z1,z2)
            classifi_loss=F.cross_entropy(logit,naics2_labels,weight=normalized_weights,label_smoothing=0.1)
            combine_loss=contrast_loss+classifi_loss
            combine_loss.backward()
            optimizer.step()
            loss_total+=combine_loss.item()
        print(f"current epoch{epoch+1},loss:{loss_total}")

    accuracy_val1,accuracy_val3,accuracy_val5,accuracy_val10=topk_accuracy(loader_val)
    results_lst.append((learning_rates,accuracy_val1))
    print(f"When learning rate is {learning_rates}, ACC@1 for validation set is {accuracy_val1}")

    if accuracy_val1>best_acc1:
        best_acc1=accuracy_val1
        torch.save(model.state_dict(),'sbert_hyperparam_best.pt')
        print(f"current best sbert model saved")

print(max(results_lst,key=lambda x: x[1]),"is the best learning rate and best ACC@1 for Validation Set")
       
```

```python
# as the models above with adamW optimizer were less effective than with SGD optimizer with lr=2e-05
# I will now do lr=1e-05 (best performing out of the 3 lrs) with SGD optimizer compare SGD optimizer lr=2e-05 (original)
pretrained=SentenceTransformer("all-MiniLM-L6-v2")
model=SBERTClassifier(pretrained,num_classes)
model.to(device)
optimizer=torch.optim.SGD(model.parameters(),lr=1e-05,momentum=0.85) #switched back to SGD

for epoch in range (5):
    model.train()
    loss_total=0

    for company_description,naics2_description, naics2_labels in loader_train: 
        naics2_labels=torch.tensor(naics2_labels).to(device)
        optimizer.zero_grad()
        z1,_,_=model(company_description)
        z2,_,logit=model(naics2_description)
        contrast_loss=loss_mnr(z1,z2)
        classifi_loss=F.cross_entropy(logit,naics2_labels,label_smoothing=0.1) # no class weights (like original) to see if it improves 
        combine_loss=contrast_loss+classifi_loss
        combine_loss.backward()
        optimizer.step()
        loss_total+=combine_loss.item()
```

```python
from sklearn.model_selection import StratifiedKFold
k_folds=StratifiedKFold(n_splits=5,shuffle=True,random_state=109)
cross_val_res=[]

train_andval['NAICS_2 Label']=encoder.transform(train_andval['NAICS_2 Title'])


for kth_fold,(idx_train,idx_val) in enumerate (k_folds.split(train_andval,train_andval['NAICS_2 Title'])):
    train_fold=train_andval.take(idx_train)
    val_fold=train_andval.take(idx_val)

    loader_train_fold=DataLoader(list(zip(train_fold['Company Description'].tolist(),train_fold['NAICS_2 Description'].tolist(),
    train_fold['NAICS_2 Label'].tolist())),batch_size=32,shuffle=True,drop_last=True)
    loader_val_fold=DataLoader(list(zip(val_fold['Company Description'].tolist(),val_fold['NAICS_2 Description'].tolist(),
    val_fold['NAICS_2 Label'].tolist())),batch_size=32,drop_last=True)

    pretrained=SentenceTransformer("all-MiniLM-L6-v2")
    model=SBERTClassifier(pretrained,num_classes)
    model.to(device)
    optimizer=torch.optim.SGD(model.parameters(),lr=2e-05,momentum=0.85)

    for epoch in range (3):
        model.train()
        loss_total=0

        for company_description,naics2_description, naics2_labels in loader_train_fold: 
            naics2_labels=torch.tensor(naics2_labels).to(device)
            optimizer.zero_grad()
            z1,_,_=model(company_description)
            z2,_,logit=model(naics2_description)
            contrast_loss=loss_mnr(z1,z2)
            classifi_loss=F.cross_entropy(logit,naics2_labels,label_smoothing=0.1)
            combine_loss=contrast_loss+classifi_loss
            combine_loss.backward()
            optimizer.step()
            loss_total+=combine_loss.item()
        
    print(f"Current fold:{kth_fold+1}")
    acc1,acc3,acc5,acc10=topk_accuracy(loader_val_fold)
    cross_val_res.append((acc1,acc3,acc5,acc10))
        
for idx, kth in enumerate([1,3,5,10]):
    avg_accuracy=sum(r[idx]for r in cross_val_res)/len(cross_val_res)
    print(f"Average ACC@{kth} for validation set = {avg_accuracy}")
```

```python
from sklearn.metrics import mean_absolute_error
import numpy as np

def topk_accuracy_with_ghg(loader):

    model.eval()
    acc1,acc3,acc5,acc10=0,0,0,0
    total=0

    ghg_per_naics2=df.groupby('NAICS_2 Label')['Carbon Intensity'].mean().to_dict()

    carbon_predict=np.array([])
    truth_ghg_company=np.array([])

    with torch.no_grad():
        for company_description, naics2_description, label, company_revenue, co2 in loader: 
            label=torch.tensor(list(label)).to(device)
            company_revenue=torch.tensor(list(company_revenue),dtype=torch.float)
            co2=torch.tensor(list(co2),dtype=torch.float)
            logit=model(company_description)[2]

            pred_top10=logit.topk(10,dim=1).indices.cpu().numpy() 

            pred_co2=torch.tensor([np.mean([ghg_per_naics2[int(naics)]for naics in top10])for top10 in pred_top10]
            ,dtype=torch.float)
            company_ghg=pred_co2*company_revenue*1000000

            carbon_predict=np.concatenate([carbon_predict,company_ghg.numpy()])
            truth_ghg_company=np.concatenate([truth_ghg_company,co2.numpy()])

            acc1+=accuracy_topk(logit,label,k=1)
            acc3+=accuracy_topk(logit,label,k=3)
            acc5+=accuracy_topk(logit,label,k=5)
            acc10+=accuracy_topk(logit,label,k=10)
            total+=len(label)
    acc_1=acc1/total
    acc_3=acc3/total
    acc_5=acc5/total
    acc_10=acc10/total
    mean_abs_error=mean_absolute_error(truth_ghg_company,carbon_predict)
    rmse=np.sqrt(np.mean(carbon_predict-truth_ghg_company)**2)
    print("Mean ABS Error for GHG avg by top 10 NAICS2 predictions :",mean_abs_error)
    print("RMSE for GHG:", rmse)
    print("ACC@1 for NAICS2 Classification:",acc_1)
    print("ACC@3 for NAICS2 Classification:",acc_3)
    print("ACC@5 for NAICS2 Classification:",acc_5)
    print("ACC@10 for NAICS2 Classification:",acc_10)

    return acc_1,acc_3,acc_5,acc_10, carbon_predict, truth_ghg_company, mean_abs_error 
```

```python
# the second ood dataset (for testing)

wiki = pd.read_csv('wikidata_global_companies_info.csv', index_col=0)
print(wiki.shape)
wiki.head(2)
```

```python
def parse_industry(s):
    if pd.isna(s): return None
    s = re.sub(r"[\[\]']", '', str(s))
    return re.sub(r'\s+', ' ', s).strip() or None

wiki['industry_clean'] = wiki['industry'].apply(parse_industry)

wiki_ood = wiki[['company name', 'description', 'industry_clean']].copy()
wiki_ood = wiki_ood.rename(columns={'company name': 'company_name', 'description': 'text', 'industry_clean': 'label'})
wiki_ood = wiki_ood.dropna(subset=['label', 'text']).reset_index(drop=True)

print(wiki_ood.shape)
print(wiki_ood['label'].value_counts().head(10))
wiki_ood.head(3)
```

```python
# cosine similarity for closest NAICS2 label mapping for the second ood dataset 
wiki_ood_embedding=model.pretrained.encode(wiki_ood['label'].tolist(),convert_to_tensor=True,device='cpu')

wiki_similar_labels=F.cosine_similarity(wiki_ood_embedding.unsqueeze(1),naics2_embedding.unsqueeze(0),dim=2)
_,wiki_idx_map=wiki_similar_labels.max(dim=1)
wiki_label_map=[encoder.classes_[ith.item()]for ith in wiki_idx_map]
```

```python
wiki_ood['label_naics2_mapped']=wiki_label_map
wiki_ood_labels=encoder.transform(wiki_ood['label_naics2_mapped'].tolist())
loader_wiki_ood=DataLoader(list(zip(wiki_ood['text'].tolist(),wiki_ood['label_naics2_mapped'].tolist(),
wiki_ood_labels.tolist())),batch_size=32,drop_last=True)
```
