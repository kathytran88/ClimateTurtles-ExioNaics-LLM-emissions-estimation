```python
import json
import glob
```

```python
result_files = sorted(glob.glob("results/session_*.json"))
print(f"Found {len(result_files)} result files\n")

all_sessions = []
for f in result_files:
    with open(f) as fp:
        all_sessions.append(json.load(fp))
```

```python
print(f"{'Session':<12} {'Strategy':<20} {'Top-1':>8} {'Top-5':>8} {'Top-10':>8} {'Macro-F1':>9} {'W-F1':>8}")
print("-" * 85)

best_session = None
best_top1 = 0

for s in all_sessions:
    top1 = s['eval_top1_accuracy']['mean']
    top5 = s['eval_top5_accuracy']['mean']
    top10 = s['eval_top10_accuracy']['mean']
    mf1 = s['eval_macro_f1']['mean']
    wf1 = s['eval_weighted_f1']['mean']
    name = s['config']['name']
    sess = s['session']

    print(f"Session {sess:<5} {name:<20} {top1:>8.4f} {top5:>8.4f} {top10:>8.4f} {mf1:>9.4f} {wf1:>8.4f}")

    if top1 > best_top1:
        best_top1 = top1
        best_session = s

print(f"\nBest session: {best_session['session']} ({best_session['config']['name']})")
print(f"  Learning rate: {best_session['config']['learning_rate']}")
print(f"  Batch size:    {best_session['config']['batch_size']}")
print(f"  Weight decay:  {best_session['config']['weight_decay']}")
print(f"  Top-1 Accuracy: {best_top1:.4f}")
```
